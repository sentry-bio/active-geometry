# Notebook B (Python mirror) — Sensitivity & Intervals

```

from decimal import Decimal, getcontext
getcontext().prec = 80
LN2 = Decimal(2).ln()
def kappa(h, n=2):
    return (Decimal(h)*LN2/(Decimal(n)-Decimal(1)))**2
def d_kappa_dh(h, n=2):
    h = Decimal(h); n=Decimal(n)
    return 2*(LN2**2)*h/((n-Decimal(1))**2)
def d_kappa_dn(h, n=2):
    h = Decimal(h); n=Decimal(n)
    return -2*(h*LN2)**2/((n-Decimal(1))**3)
h0, n0 = Decimal('1.6'), Decimal('2')
print('kappa(h0,n0)=', +kappa(h0,n0))
print('∂κ/∂h=', +d_kappa_dh(h0,n0))
print('∂κ/∂n=', +d_kappa_dn(h0,n0))
σh, σn = Decimal('0.10'), Decimal('0.06')
σk = ((d_kappa_dh(h0,n0)*σh)**2 + (d_kappa_dn(h0,n0)*σn)**2).sqrt()
print('Propagated σ_k=', +σk)
print('κ interval for h∈[1.50,1.70]:', +kappa('1.50'), +kappa('1.70'))
```
